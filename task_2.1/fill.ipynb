```python
import pandas as pd
import nltk
nltk.download('punkt_tab')
```

```python
df = pd.read_csv('human_essays_v1.5.csv')
df.head()
```

```python
df1 = df[df["introduction"].isna() | (df["introduction"] == "")]
len(df1)
```

```python
for index, row in df1.iterrows():
  sentences = sentences = nltk.sent_tokenize(row["essays"])
  intro = sentences[:3]
  text = " ".join(intro)
  df1.at[index, "introduction"] = text

df1.to_csv("human_essays_v1.6_fill_blank.csv", index=False)
```
